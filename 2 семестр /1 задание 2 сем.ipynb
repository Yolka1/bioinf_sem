Напишите функцию, которая будет возвращать индекс минимального элемента в списке. Если их несколько, то вернет все минимальные положения.

```python
def find_min_indexes(lst):
    # Если список пустой, возвращаем пустой список индексов
    if not lst:
        return []
    
    # 1. Принимаем самый первый элемент за минимальный
    current_min = lst[0]
    
    # Сюда будем складывать индексы. Первый индекс (0) уже подходит.
    min_indexes = [0]
    
    # 2. Перебираем список, начиная со второго элемента (индекс 1)
    # Используем range(len(lst)), чтобы сразу видеть и индекс (i), и само число lst[i]
    for i in range(1, len(lst)):
        if lst[i] < current_min:
            # Нашли число МЕНЬШЕ текущего минимума!
            current_min = lst[i]  # Обновляем минимум
            min_indexes = [i]     # Сбрасываем старые индексы и записываем новый
            
        elif lst[i] == current_min:
            # Нашли число, РАВНОЕ текущему минимуму!
            min_indexes.append(i) # Просто дописываем индекс в список
            
    return min_indexes

# --- Блок проверки (с генерацией случайного списка) ---
import random

# Создаем случайный список из 10 чисел от 1 до 5 (чтобы точно были повторения)
random_list = [random.randint(1, 5) for _ in range(10)]

print("Сгенерированный список:", random_list)
print("Индексы минимальных элементов:", find_min_indexes(random_list))
```

Напишите функцию, которая возвращает отсортированный методом пузырька список. Для продвинутых: попробуйте написать декоратор, который будет отслеживать выполнение этой функции.

```python
import random

# 1. Создаем функцию для сортировки
def bubble_sort(my_list):
    # Узнаем длину списка (сколько в нем элементов)
    n = len(my_list)
    
    # Главный цикл: он отвечает за то, сколько раз мы пройдемся по списку.
    # Чтобы полностью отсортировать список, нужно пройтись по нему n раз.
    for i in range(n):
        
        # Внутренний цикл: идет по элементам от начала до предпоследнего.
        # Мы не идем до самого конца, чтобы не выйти за границы списка при сравнении (j + 1).
        for j in range(0, n - 1):
            
            # Сравниваем два соседних числа: текущее my_list[j] и следующее my_list[j + 1]
            if my_list[j] > my_list[j + 1]:
                
                # Если левое число больше правого, меняем их местами!
                # В Питон это делается одной простой строчкой:
                my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]
                
    # Возвращаем уже красивый, отсортированный список
    return my_list


# --- Блок проверки (создаем случайный список и сортируем его) ---

# Генерируем список из 7 случайных чисел от 1 to 50
random_list = [random.randint(1, 50) for _ in range(7)]

print("Оригинальный список:  ", random_list)

# Вызываем нашу функцию и сохраняем результат в новую переменную
sorted_result = bubble_sort(random_list)

print("Отсортированный список:", sorted_result)
```

Напишите функцию-генератор случайной нуклеотидной последовательности заданной пользователем длины.

```python
import random

# Создаем функцию, которая принимает нужную длину (length)
def generate_dna(length):
    # Наша "библиотека" нуклеотидов. 
    # Вместо сложного словаря можно взять просто строку, где буквы идут под номерами 0, 1, 2, 3
    nucleotides = "ATGC"
    
    # Сюда мы будем по одной добавлять наши случайные буквы
    result_sequence = ""
    
    # Запускаем цикл столько раз, какую длину попросил пользователь
    for i in range(length):
        # Генерируем случайное число от 0 до 3 (всего 4 варианта)
        random_number = random.randint(0, 3)
        
        # Берем букву из нашей "библиотеки" по этому случайному номеру
        random_letter = nucleotides[random_number]
        
        # Добавляем букву к нашей итоговой строке
        result_sequence = result_sequence + random_letter
        
    # Возвращаем готовую последовательность
    return result_sequence


# --- Блок проверки программы ---

# Запрашиваем длину у пользователя (например, 15)
user_length = int(input("Введите желаемую длину последовательности: "))

# Вызываем функцию и печатаем результат
dna_result = generate_dna(user_length)
print("Ваша случайная ДНК-последовательность:")
print(dna_result)
```

Угадай число

```python
import random

# Создаем функцию для игры. Ей не нужны входные параметры, 
# потому что всё происходит внутри неё.
def guess_the_number_game():
    # 1. Компьютер загадывает случайное число от 1 до 100
    secret_number = random.randint(1, 100)
    
    print("Я загадал число от 1 до 100. У тебя есть 10 попыток, чтобы угадать его!")
    
    # 2. Запускаем цикл для 10 попыток (от 1 до 10 включительно)
    for attempt in range(1, 11):
        # Спрашиваем число у пользователя
        user_guess = int(input(f"Попытка {attempt}. Введи твое число: "))
        
        # 3. Сравниваем введенное число с загаданным
        if user_guess == secret_number:
            print(f"Поздравляю! Ты угадал число за {attempt} попыток!")
            return  # Выходим из функции, так как игра завершена победой
            
        elif user_guess < secret_number:
            print("Загаданное число БОЛЬШЕ твоего.")
            
        else:
            print("Загаданное число МЕНЬШЕ твоего.")
            
    # Если цикл закончился (10 попыток прошли), а `return` не сработал — игрок проиграл
    print(f"Попытки закончились! К сожалению, ты проиграл. Я загадал число {secret_number}.")


# --- Вызов функции (запуск игры) ---
# Просто пишем имя функции со скобками, и игра начинается!
guess_the_number_game()
```
